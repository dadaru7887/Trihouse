### 베터리 상태 체크 패키지

```python
import rclpy

from trihouse_interfaces.msg import BatteryPolicyState
from trihouse_pinky_fleet.status_node import StatusNode


def test_robot_status_uses_battery_policy_message():
    if not rclpy.ok():
        rclpy.init()

    node = StatusNode()

    try:
        policy = BatteryPolicyState()
        policy.robot_id = "PK-01"
        policy.mode = BatteryPolicyState.MODE_RETURN_TO_CHARGE
        policy.battery_percentage = 18.0
        policy.reason = "battery below return threshold"

        node.battery = 18.0             # 측정된 단순 배터리 잔량
        node.battery_policy = policy

        message = node._build_message()

        assert message.battery_percentage == 18.0
        assert isinstance(message.battery_policy, BatteryPolicyState)
        assert message.battery_policy.mode == BatteryPolicyState.MODE_RETURN_TO_CHARGE
        assert message.battery_policy.reason == "battery below return threshold"

    finally:
        node.destroy_node()
        if rclpy.ok():
            rclpy.shutdown()


test_robot_status_uses_battery_policy_message()
```
